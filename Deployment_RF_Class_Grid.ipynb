```python
import pickle
loaded_model=pickle.load(open("CKD_Prediction_RF_Classification.sav", 'rb'))
```

```python
age = int(input("Please enter the age: "))
bp = float(input("Please enter the bp value: "))
al = float(input("Please enter the al value: "))
su = float(input("Please enter the su value: "))
sgr = float(input("Please enter the sgr value: "))
bu = float(input("Please enter the bu value: "))
sc = float(input("Please enter the sc value: "))
sod = float(input("Please enter the sod value: "))
pot = float(input("Please enter the pot value: "))
hrmo = float(input("Please enter the hrmo value: "))
pcv = float(input("Please enter the pcv value: "))
wc = float(input("Please enter the wc value: "))
rc = float(input("Please enter the rc value: "))
sg_type = input("Please enter your sg type: ")
sg_b = 0
sg_c = 0
sg_d = 0
sg_e = 0
if sg_type == "b":
    sg_b = 1
if sg_type == "c":
     sg_c = 1
if sg_type == "d":
     sg_d = 1
if sg_type == "e":
     sg_e = 1
rbc = int(input("If the rbc count is normal, please type 1 else type 0: "))
pc = int(input("If the pc count is normal, please type 1 else type 0: "))
pcc = int(input("If pcc is present, please type 1 else type 0: "))
ba = int(input("If ba is present, please type 1 else type 0: "))
htn = int(input("If patient has hypertension, please type 1 else type 0: "))
dm = int(input("If dm is positive, please type 1 else type 0: "))
cad = int(input("If cad is positive, please type 1 else type 0: "))
appet = int(input("If appet is positive, please type 1 else type 0: "))
pe = int(input("If pe is positive, please type 1 else type 0: "))
ane = int(input("If patient is anemic, please type 1 else type 0: "))
```

```python
final_result = loaded_model.predict([[age, bp, al, su, sgr, bu, sc, sod, pot, hrmo, pcv, wc, rc, sg_b, sg_c, sg_d, sg_e, rbc, pc, pcc, ba, htn, dm, cad, appet, pe, ane]])
```

```python
if final_result == 1:
    print("The patient is diagnosed with with CKD")
else:
    print("CKD is 'Negative' for the entered patient's record")
```
