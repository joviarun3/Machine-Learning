```python
import pandas as pd
dataset=pd.read_csv("CKD.csv")
dataset=pd.get_dummies(dataset, drop_first = True, dtype =int)
dataset
```

```python
dataset.columns
```

```python
ind=dataset[['age', 'bp', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hrmo', 'pcv',
       'wc', 'rc', 'sg_b', 'sg_c', 'sg_d', 'sg_e', 'rbc_normal', 'pc_normal',
       'pcc_present', 'ba_present', 'htn_yes', 'dm_yes', 'cad_yes',
       'appet_yes', 'pe_yes', 'ane_yes']]
dep = dataset[['classification_yes']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(ind, dep, test_size=0.3, random_state=0)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import KFold
from sklearn.tree import DecisionTreeClassifier

param_grid = {'criterion':['gini','entropy'],
              'max_features': [1, 'sqrt','log2'],
              'splitter':['best','random']}

grid = GridSearchCV(DecisionTreeClassifier(), param_grid, refit = True, verbose = 3,n_jobs=-1, cv=KFold(), scoring='f1_weighted')

grid.fit(X_train, y_train)
```

```python
re=grid.cv_results_
table=pd.DataFrame.from_dict(re)
table.to_excel("DT_Class_GRID.xlsx", index=False)
table
```

```python
grid_pred = grid.predict(X_test) 
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, grid_pred)
#print(cm)
```

```python
from sklearn.metrics import classification_report
clf_report = classification_report(y_test, grid_pred)
#print(clf_report)
```

```python
from sklearn.metrics import roc_auc_score
ras = roc_auc_score(y_test,grid.predict_proba(X_test)[:,1])
#Receiver-operating characteristic curve (ROC) Area under the curve (AUC)
#This roc_auc_score needs probability as one of the inputs. Sowe used predict_proba fuction (from GridSearchCV)
#[:,1] this is just which rows and colums have this probaility values. Here..[:,1] mean..each rows and first collumn.
```

```python
print("Evaluation metrics details for DT are given below")
print("1) Confusion Matrix:\n",cm)
print('2) classification report:\n', clf_report)
print("3) roc_auc_score is", ras)
print("4) The best parameters from GridSearchCV for the created DT model are : \n" , grid.best_params_)
```
