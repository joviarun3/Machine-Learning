```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("Mall_Customers.csv")
dataset
```

```python
X = dataset.iloc[ :, [3,4]].values
#Check if X contains NaN, infinite, or missing values. Spectral Clustering doesn’t handle these.
print(np.any(np.isnan(X)))  # Should return False
print(np.any(np.isinf(X)))  # Should return False
```

```python
#Compute silhouette scores for a range of cluster numbers and choose the one with the highest score.
from sklearn.cluster import SpectralClustering
from sklearn.metrics import silhouette_score
sil_scores = []
for k in range(2, 11):  # Silhouette score requires at least 2 clusters
    clustering = SpectralClustering(n_clusters=k, random_state=42, affinity='nearest_neighbors')
    clustering.fit(X)
    labels = clustering.labels_
    sil_score = silhouette_score(X, labels)
    sil_scores.append(sil_score)

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))
plt.plot(range(2, 11), sil_scores, marker='o', color='orange')
plt.title('Silhouette Score Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.grid()
plt.show()
```

```python
#I am choosing 5 as number of clusters. because silhoutte score is high for when number of cluster is 5
from sklearn.cluster import SpectralClustering
clustering = SpectralClustering(n_clusters=5, random_state=42, affinity='nearest_neighbors')
y = clustering.fit_predict(X)
```

```python
y
```

```python
supervised = dataset
supervised['Cluster_group']=y
supervised.to_csv("Spectral_Clusters.csv")
```
